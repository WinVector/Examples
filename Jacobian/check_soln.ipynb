Example of a non-trivial invertible polynomial map.

```python
import sympy as sp
```

```python
z_1, z_2, z_3 = sp.symbols('z_1, z_2, z_3')
```

```python
g1, g2, g3 = (
    z_1 + 5*z_2**2 - 5*z_3, 
    z_2 + z_3**2, 
    z_1/5 + z_2**2
    )
```

```python
g1
```

```python
g2
```

```python
g3
```

Check determinant is 1 everywhere.

```python
def compute_det(F, *, variables):
    # Vector function and its Jacobian
    J = sp.Matrix(F).jacobian(variables)
    # Determinant of the Jacobian
    detJ = sp.simplify(J.det())
    return detJ
```

```python
compute_det((g1, g2, g3), variables=(z_1, z_2, z_3))
```

Check is invertible.

```python
y_1, y_2, y_3 = sp.symbols('y_1, y_2, y_3')
eqns = [
    g1 - y_1, g2 - y_2, g3 - y_3
]
soln = sp.solve(eqns, (z_1, z_2, z_3))
assert len(soln) == 1
h1, h2, h3 = soln[0]
```

Bring in Levent Alpöge's counter-example (re-arranged and re-scaled to have det 1).

```python

f1 = y_2 + 3 * y_1 * (1 + y_1 * y_2)**2 * y_3 + 3 * y_1 * y_2**2 * (4 + 3 * y_1* y_2)
f2 = (1 + y_1 * y_2)**3 * y_3 + y_2**2 * (1 + y_1 * y_2) * (4 + 3* y_1 * y_2)
f3 =  (2 * y_1 - 3 * y_1**2 * y_2 - y_1**3 * y_3) / 2
```

Check determinant.

```python
compute_det((f1, f2, f3), variables=(y_1, y_2, y_3))
```

Exhibit not 1 to 1 (so not invertible).

```python
p1, p2 = (-3, sp.Rational(1, 3), 1), (1, sp.Rational(-2, 3), -8)
```

```python
p1
```

```python
p2
```

```python
def map_f(fn, *, point, variables):
    """Evaluate polynomial map"""
    return tuple(fi.subs({zi: pi for zi, pi in zip(variables, point)}) for fi in fn)
```

```python
map_f((f1, f2, f3), point=p1, variables=(y_1, y_2, y_3))
```

```python
map_f((f1, f2, f3), point=p2, variables=(y_1, y_2, y_3))
```
